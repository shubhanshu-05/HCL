# Electricity Consumption Data Analysis

This notebook analyzes `electricity_data.csv`, cleans the data, creates useful derived features, generates visualizations, and summarizes factors associated with monthly electricity consumption.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('electricity_processed.csv')
df.head()
```

## 1. Dataset Overview

The dataset contains household-level electricity information such as city, home type, cooling type, billing plan, household size, floor area, temperature, appliance count, building age, reading date, and monthly electricity consumption.

```python
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
print('\nMissing values:\n', df.isna().sum())
print('\nData types:\n', df.dtypes)
```

## 2. Data Cleaning and Processing

- Removed duplicate rows.
- Standardized city names and other text fields.
- Converted `Reading_Date` to datetime.
- Treated non-positive household size as invalid and replaced it using the median.
- Filled missing categorical data using the mode and missing numerical data using the median.
- Added year, month, consumption bands, and units consumed per person.

```python
df.describe(include='all').T
```

## 3. Descriptive Statistics

The main target variable is `Monthly_Units_Consumed`. Use the statistics below to understand the central tendency and spread of household electricity usage.

```python
df[['Household_Size','Floor_Area_Sqft','Avg_Temp_C','Appliance_Count',
    'Building_Age_Years','Monthly_Units_Consumed','Units_Per_Person']].describe()
```

## 4. Distribution of Electricity Consumption

This histogram shows how monthly electricity usage is distributed across households. It helps identify whether consumption is concentrated around a typical range or whether there are many high-usage households.

```python
plt.figure(figsize=(8,5))
plt.hist(df['Monthly_Units_Consumed'], bins=12, edgecolor='white')
plt.title('Distribution of Monthly Electricity Consumption')
plt.xlabel('Monthly Units Consumed')
plt.ylabel('Households')
plt.show()
```

## 5. Consumption Bands

Households are grouped into Low, Moderate, High, and Very High consumption bands based on monthly units consumed.

```python
df['Consumption_Band'].value_counts().reindex(['Low','Moderate','High','Very High'])
```

## 6. Temperature and Consumption

The scatter plot examines whether warmer average temperatures are associated with greater electricity consumption.

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Avg_Temp_C'], df['Monthly_Units_Consumed'], alpha=.55)
plt.title('Average Temperature vs Monthly Electricity Consumption')
plt.xlabel('Average Temperature (°C)')
plt.ylabel('Monthly Units Consumed')
plt.grid(alpha=.25)
plt.show()
```

## 7. Floor Area and Consumption

This plot checks whether larger homes tend to consume more electricity.

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Floor_Area_Sqft'], df['Monthly_Units_Consumed'], alpha=.55)
plt.title('Floor Area vs Monthly Electricity Consumption')
plt.xlabel('Floor Area (sq ft)')
plt.ylabel('Monthly Units Consumed')
plt.grid(alpha=.25)
plt.show()
```

## 8. Correlation Analysis

The correlation matrix compares numeric variables. Values closer to +1 indicate a stronger positive relationship, values near 0 indicate a weak linear relationship, and negative values indicate an inverse relationship.

```python
num_cols = ['Household_Size','Floor_Area_Sqft','Avg_Temp_C','Appliance_Count',
            'Building_Age_Years','Monthly_Units_Consumed','Units_Per_Person']
df[num_cols].corr()['Monthly_Units_Consumed'].sort_values(ascending=False)
```

## 9. City-wise Analysis

Comparing average monthly consumption across cities helps identify geographic differences in electricity usage.

```python
df.groupby('City')['Monthly_Units_Consumed'].agg(['count','mean','median']).sort_values('mean', ascending=False)
```

## 10. Cooling Type Analysis

Cooling equipment can be an important contributor to household electricity demand. This comparison shows average usage by cooling type.

```python
df.groupby('Cooling_Type')['Monthly_Units_Consumed'].agg(['count','mean','median']).sort_values('mean', ascending=False)
```

## 11. Home Type Analysis

This comparison examines electricity usage across apartments, independent houses, and studios.

```python
df.groupby('Home_Type')['Monthly_Units_Consumed'].agg(['count','mean','median']).sort_values('mean', ascending=False)
```

## 12. Key Insights

1. The dataset contains 255 recorded household observations before duplicate removal and cleaning.
2. Consumption varies substantially between households, so the consumption bands help separate typical and high-use homes.
3. Temperature, floor area, appliance count, and household size can be compared with monthly consumption using the correlation matrix and scatter plots.
4. City, cooling type, and home type provide useful categorical comparisons for understanding differences in electricity demand.
5. The processed dataset is suitable for further modeling, such as predicting monthly electricity consumption.

## 13. Conclusion

The analysis provides a cleaned and structured view of household electricity consumption. The strongest relationships in the numeric data can be identified from the correlation table, while the categorical comparisons show how consumption differs by location, cooling system, and type of home. The processed CSV can be used for further visualization or machine-learning work.
